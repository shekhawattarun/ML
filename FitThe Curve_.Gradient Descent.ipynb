```python
#Ans3:- Code to fit the curve
import numpy as np
import matplotlib.pyplot as plt

# Input and target data
x = np.array([-1, -0.8, -0.6, -0.4, -0.2, 0, 0.2, 0.4, 0.6, 0.8, 1])
t = np.array([-4.9, -3.5, -2.8, 0.8, 0.3, -1.6, -1.3, 0.5, 2.1, 2.9, 5.6])

mu = np.array([-1, -0.33, 0.33, 1])
sigma = 1.0

def gaussian_basis(x, mu, sigma):
    return np.exp(-0.5 * ((x - mu) / sigma) ** 2)

Phi = np.array([[gaussian_basis(xi, muj, sigma) for muj in mu] for xi in x])

#weights using the least squares solution
w = np.linalg.inv(Phi.T @ Phi) @ Phi.T @ t

#generate predictions
x_fit = np.linspace(-1, 1, 100)
Phi_fit = np.array([[gaussian_basis(xi, muj, sigma) for muj in mu] for xi in x_fit])
t_fit = Phi_fit @ w

#original data and the fitted curve
plt.scatter(x, t, color='red', label='Data Points')
plt.plot(x_fit, t_fit, label='Fitted Curve')
plt.xlabel('x')
plt.ylabel('t')
plt.legend()
plt.show()
```

```python
#Ans2:Code for Gradient Descent:
import numpy as np

# Initial point
x = np.array([1.0, 2.0, 3.0])

# Learning rate
eta = 0.1

# Perform 3 iterations of gradient descent
for i in range(3):
    gradient = x  # Since ∇f(x) = x for this function
    x = x - eta * gradient  # Update rule
    print(f"Iteration {i+1}: x = {x}, f(x) = {0.5 * np.sum(x**2)}")
```
